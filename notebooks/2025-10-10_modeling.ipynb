#### The cost of a taxi ride

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Подготовка данных (наши наблюдения)
# X - это входные данные (километры).
X = np.array([[1], [5], [10]]) 

# y - это ответы (цена). 
y = np.array([150, 350, 600])

# 2. Создание модели
model = LinearRegression()

# 3. Обучение 
# Команда .fit() запускает процесс подбора весов (w и b).
# Именно здесь модель "смотрит" на данные и ищет идеальную линию.
model.fit(X, y)

# 4. Проверка результатов
print(f"Модель обучена!")
print(f"Найденный вес (w - цена за км): {model.coef_[0]:.2f}")
print(f"Найденный сдвиг (b - цена посадки): {model.intercept_:.2f}")

# 5. Предсказание
# Мы хотели узнать, сколько стоит поездка на 20 км.
km_new = [[20]] 
prediction = model.predict(km_new)

print(f"Прогноз стоимости для 20 км: {prediction[0]:.2f} рублей")
```

```python
import matplotlib.pyplot as plt

# Рисуем точки (реальные данные)
plt.scatter(X, y, color='blue', label='Реальные поездки')

# Рисуем линию
# Просим предсказать цены для всех наших километров
y_pred = model.predict(X) 
plt.plot(X, y_pred, color='red', label='Линия регрессии')

plt.xlabel('Километры')
plt.ylabel('Цена (руб)')
plt.legend()
plt.show()
```
